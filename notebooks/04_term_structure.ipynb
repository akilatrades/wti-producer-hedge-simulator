# 04 — WTI Term Structure
Business question: how does the relationship between nearby and deferred WTI futures change through time?


## Load the saved C1-C4 history and calculate curve features

```python
import matplotlib.pyplot as plt
import pandas as pd

from src.term_structure import (
    add_curve_features,
    add_curve_metrics,
    summarize_curve_regimes,
)

curve = pd.read_csv(
    "../data/eia_wti_curve_monthly_2015_2024.csv",
    parse_dates=["date"],
    index_col="date",
)

curve = add_curve_metrics(curve)
curve = add_curve_features(curve, zscore_window=24)

summarize_curve_regimes(curve)
```

## Plot C1-C4

```python
ax = curve["c1_c4_spread"].plot(figsize=(11, 5))
ax.axhline(0, linewidth=1)
ax.set_title("WTI C1-C4 Spread")
ax.set_ylabel("$/bbl")
plt.show()
```

## Review unusual curve observations

```python
curve[
    [
        "c1_c4_spread",
        "curve_curvature",
        "c1_c4_zscore",
        "c1_c4_percentile",
        "curve_regime",
    ]
].tail(12)
```

## Interpretation

Backwardation, contango, slope, and z-score describe the curve. They are market-state variables, not automatic directional trading signals.
