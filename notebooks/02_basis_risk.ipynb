Business question: what risk remains when physical crude is priced at Midland but the hedge is linked to Cushing WTI?


## Run the scenario grid

```python
from src.basis_risk import compare_basis_scenarios

basis = compare_basis_scenarios()
basis.head()
```

## Focus on residual basis risk

```python
basis[
    [
        "basis_hedge_ratio",
        "realized_midland_basis",
        "residual_revenue_risk",
    ]
].pivot(
    index="realized_midland_basis",
    columns="basis_hedge_ratio",
    values="residual_revenue_risk",
)
```

## Interpretation

A Cushing futures hedge can reduce the broad WTI price move while leaving a Midland/Cushing location mismatch. That is why basis risk should be measured separately.
