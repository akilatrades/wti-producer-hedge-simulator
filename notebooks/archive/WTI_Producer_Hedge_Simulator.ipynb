## WTI futures curve / term structure

WTI futures trade across multiple delivery periods at the same time. Those prices, arranged from nearer delivery to later delivery, form the **futures curve**.

For this analysis:

```
C1 = nearest delivery position in the historical EIA series
C2 = next delivery position
C3 = next delivery position
C4 = fourth delivery position
```

C1 is not one permanent contract. It means the nearest position at each point in the historical data, so the exact contract represented by C1 changes through time.

The saved curve data is in:

`data/eia_wti_curve_monthly_2015_2024.csv`

````python
curve = pd.read_csv(
    ROOT / "data" / "eia_wti_curve_monthly_2015_2024.csv",
    parse_dates=["date"],
    index_col="date",
)

curve.head()
````

### Compare one observation before calculating anything

Each row contains C1, C2, C3, and C4 for the same monthly observation.

If one row looks like:

```
C1 = 80
C2 = 79
C3 = 78
C4 = 77
```

the nearer delivery positions are priced above the later delivery positions.

The next step turns that relationship into a number.

````python
example = curve.iloc[0]

print("Date:", curve.index[0].date())
print("C1:", example["contract_1"])
print("C2:", example["contract_2"])
print("C3:", example["contract_3"])
print("C4:", example["contract_4"])
````

### Calculate calendar spreads

A calendar spread is the price difference between two delivery positions.

The project calculates:

```
C1-C2 = C1 price - C2 price
C1-C3 = C1 price - C3 price
C1-C4 = C1 price - C4 price
```

The main measure used here is C1-C4.

A positive C1-C4 number means C1 is above C4.  
A negative C1-C4 number means C1 is below C4.

````python
c1 = example["contract_1"]
c4 = example["contract_4"]
c1_c4 = c1 - c4

print("C1 - C4 =", round(c1_c4, 2))
````

### Classify the curve shape

The project uses this rule:

```
C1-C4 > +0.25 -> backwardation
C1-C4 < -0.25 -> contango
otherwise      -> flat
```

**Backwardation** means nearby futures are above later futures.

**Contango** means later futures are above nearby futures.

**Flat** means the difference is small under the project's chosen threshold.

The 0.25 cutoff is a project setting, not a universal market rule. These labels describe the relationship between delivery periods; they do not guarantee where WTI will move next.

````python
curve_with_metrics = add_curve_metrics(
    curve,
    flat_threshold=0.25,
)

curve_with_metrics[
    [
        "contract_1",
        "contract_2",
        "contract_3",
        "contract_4",
        "c1_c2_spread",
        "c1_c3_spread",
        "c1_c4_spread",
        "curve_regime",
    ]
].head()
````

### Summarize the historical curve

Once every observation has a curve label, the project counts how often each shape appeared and calculates the average C1-C4 spread and average C1 price for each group.

This is descriptive analysis. It shows what the historical sample looked like; it does not rank one curve shape as better than another.

````python
curve_summary = summarize_curve_regimes(curve_with_metrics)

curve_summary
````

### Why the curve matters to the producer

A producer may have expected production in January, February, March, April, and later months. Those barrels do not all arrive at the same time.

That means a more realistic hedge process eventually needs to consider **which futures delivery period best matches each physical production period**.

The curve analysis adds that time dimension to the project. It helps answer questions such as:

- How different are nearby and later WTI prices?
- Which contract month is closer to the timing of the physical exposure?
- What happens if the curve changes shape?
- What additional risk appears when a hedge is rolled into a later contract?

The current project does not yet build a full contract-by-contract hedge schedule. This analysis provides the price structure needed to add that later.
