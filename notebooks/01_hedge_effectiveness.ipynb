# 01 — Hedge Effectiveness
Business question: how much does a short WTI futures hedge reduce the producer's monthly revenue variability?


## Load saved historical results

The project keeps generated results in `outputs/` so the analysis can be reviewed without rerunning the market-data download.

```python
import matplotlib.pyplot as plt
import pandas as pd

summary = pd.read_csv("../outputs/hedge_ratio_summary.csv")
monthly = pd.read_csv("../outputs/monthly_analysis.csv")

summary
```

## Compare hedge effectiveness

Higher hedge effectiveness means the modeled hedge reduced more revenue-surprise variance relative to the unhedged case.

```python
ax = summary.plot(
    x="hedge_ratio",
    y="hedge_effectiveness",
    marker="o",
    legend=False,
)
ax.set_title("Hedge Effectiveness by Hedge Ratio")
ax.set_xlabel("Hedge Ratio")
ax.set_ylabel("Variance Reduction")
plt.show()
```

## Interpretation

The historical result is a risk-reduction benchmark, not a recommendation to maximize the hedge ratio. Commercial policy also depends on production uncertainty, basis exposure, liquidity, accounting treatment, and hedge limits.
